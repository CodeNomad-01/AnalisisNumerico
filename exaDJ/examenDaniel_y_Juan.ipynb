```python
import matplotlib.pyplot as plt
import modelos as ml
import numpy as np

#td
t = np.array([0, 0.5, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5], float)
#nd
r = np.array([1, 0.994, 0.990, 0.985, 0.979, 0.977, 0.972, 0.969, 0.967, 0.960, 0.956, 0.952], float)

#grafico de dispercion
plt.figure(figsize=(7,4))
plt.plot(t, r, 'bo')
plt.grid()
plt.show()

ml.escalas(t, r)


```

```python
b1 = ml.coeficiente_determinacion(t, np.log(r)) 
c1 = ml.coeficiente_determinacion(t, np.sqrt(r)) 

print(b1)
print(c1)

m, b = ml.minimos_cuadrados(t, np.log(r))

print()
print(f'Pendiente m = {m:.4f}')
print(f'Corte    b = {b:.4f}')
print()
print(f'Recta estirada:  log(N) = {m:.4f} * t + {b:.4f}')

t_med = np.log(0.5) / m

print(f'Vida media = {t_med:.4f} años')

modelo = lambda t: np.exp(b) * np.exp(m * t)

print(f'Modelo:   N(t) = {np.exp(b)} * exp({m} * t)')
print()
print(f'Prediccion a t=7 años:   {modelo(7)}')
print(f'Prediccion a t=10 años:   {modelo(10)}')
print((modelo(10))*100, "% original tras 10 años")
```
